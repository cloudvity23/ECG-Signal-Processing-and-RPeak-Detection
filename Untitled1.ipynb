```python
# ============================================================
# FINAL ECG R-PEAK DETECTION PIPELINE
# DWT (SWT) + Random Forest + Refractory
# Tolerance-based evaluation & plotting
# ============================================================

import numpy as np
import wfdb
import pywt
import matplotlib.pyplot as plt
import pandas as pd

from scipy.signal import find_peaks
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, recall_score, precision_score,
    f1_score, confusion_matrix, ConfusionMatrixDisplay
)

# ===================== CONFIG =====================
DATA_FOLDER = r"C:\Users\risha\Downloads\mit-bih-arrhythmia-database-1.0.0\mit-bih-arrhythmia-database-1.0.0"
FS = 360
TOL_SEC = 0.05          # 50 ms tolerance
REFRACTORY_SEC = 0.25
THRESHOLD = 0.48

ALL_RECORDS = [
    '100','101','102','103','104','105','106','107','108','109',
    '111','112','113','114','115','116','117','118','119',
    '121','122','123','124',
    '200','201','202','203','205','207','208','209','210',
    '212','213','214','215','217','219','220','221','222','223',
    '228','230','231','232','233','234'
]

# ===================== DWT ENERGY =====================
def dwt_energy_signal(ecg):
    swt = pywt.swt(ecg, 'sym4', level=4)
    filtered = []

    for i, (cA, cD) in enumerate(swt, start=1):
        if i in [3, 4]:
            filtered.append((np.zeros_like(cA), cD))
        else:
            filtered.append((np.zeros_like(cA), np.zeros_like(cD)))

    rec = pywt.iswt(filtered, 'sym4')
    return rec, rec**2

# ===================== CANDIDATE PEAKS =====================
def detect_candidates(ecg):
    _, energy = dwt_energy_signal(ecg)
    thresh = 6 * np.mean(energy)
    peaks, _ = find_peaks(
        energy,
        height=thresh,
        distance=int(0.20 * FS)
    )
    return peaks

# ===================== REFRACTORY FILTER =====================
def apply_refractory(peaks, proba):
    keep = []
    last = -np.inf
    refr = int(REFRACTORY_SEC * FS)

    for p, pr in sorted(zip(peaks, proba)):
        if p - last > refr:
            keep.append(p)
            last = p
        else:
            if pr > proba[list(peaks).index(last)]:
                keep[-1] = p
                last = p

    return np.array(keep)

# ===================== DATASET BUILDER =====================
def build_dataset(record):
    rec = wfdb.rdrecord(f"{DATA_FOLDER}/{record}")
    ecg = rec.p_signal[:, 0]

    ann = wfdb.rdann(f"{DATA_FOLDER}/{record}", 'atr')
    ann_peaks = ann.sample

    candidates = detect_candidates(ecg)

    X, y, locs = [], [], []
    tol = int(TOL_SEC * FS)

    for p in candidates:
        if p < 10 or p > len(ecg) - 10:
            continue

        win = ecg[p-10:p+10]
        qrs_width = np.sum(np.abs(np.diff(win)) > 0.05)

        feats = [
            np.max(win), np.min(win), np.mean(win), np.std(win),
            np.sum(np.abs(win)), win[10], np.max(np.diff(win)),
            np.var(win), np.sqrt(np.mean(win**2)), qrs_width
        ]

        X.append(feats)
        locs.append(p)
        y.append(1 if np.any(np.abs(ann_peaks - p) <= tol) else 0)

    return np.array(X), np.array(y), np.array(locs), ecg, ann_peaks

# ===================== PEAK MATCHING (CRITICAL) =====================
def match_peaks(detected, annotated):
    tol = int(TOL_SEC * FS)
    matches = []

    for d in detected:
        diffs = np.abs(annotated - d)
        idx = np.argmin(diffs)
        if diffs[idx] <= tol:
            matches.append((d, annotated[idx]))

    return np.array(matches)

# ===================== FINAL PEAKS =====================
def get_final_peaks(record, rf):
    X, y, locs, ecg, ann = build_dataset(record)
    proba = rf.predict_proba(X)[:, 1]
    y_pred = (proba >= THRESHOLD).astype(int)

    idx = np.where(y_pred == 1)[0]
    final_peaks = apply_refractory(locs[idx], proba[idx])

    return ecg, final_peaks, ann

# ===================== CORRECT MATCHED PLOT =====================
def plot_matched_detected_vs_annotated(ecg, detected, annotated,
                                       start_s=20, duration_s=10):
    matches = match_peaks(detected, annotated)

    start = int(start_s * FS)
    end = int((start_s + duration_s) * FS)
    t = np.arange(start, end) / FS

    plt.figure(figsize=(14, 4))
    plt.plot(t, ecg[start:end], color='black', linewidth=0.8)

    for d, a in matches:
        if start <= d < end:
            plt.plot(d/FS, ecg[d], 'ro', markersize=6)
            plt.plot(a/FS, ecg[a], 'gx', markersize=8)
            plt.plot([d/FS, a/FS], [ecg[d], ecg[a]], 'gray', linestyle='--')

    plt.title("Matched Detected vs Annotated R-peaks (Tolerance-based)")
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude (mV)")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

# ===================== TRAIN / TEST SPLIT =====================
train_records, test_records = train_test_split(
    ALL_RECORDS, test_size=0.20, random_state=42
)

# ===================== TRAIN RF =====================
X_train, y_train = [], []

for r in train_records:
    X, y, _, _, _ = build_dataset(r)
    X_train.append(X)
    y_train.append(y)

X_train = np.vstack(X_train)
y_train = np.hstack(y_train)

rf = RandomForestClassifier(
    n_estimators=400,
    max_depth=22,
    min_samples_leaf=3,
    min_samples_split=6,
    class_weight={0:1, 1:2.5},
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

# ===================== EVALUATION =====================
rows, all_true, all_pred = [], [], []

for r in test_records:
    X, y, locs, _, _ = build_dataset(r)
    proba = rf.predict_proba(X)[:, 1]
    y_pred = (proba >= THRESHOLD).astype(int)

    idx = np.where(y_pred == 1)[0]
    final_peaks = apply_refractory(locs[idx], proba[idx])

    final_labels = np.array([
        1 if any(abs(final_peaks - p) <= int(TOL_SEC*FS)) else 0
        for p in locs
    ])

    rows.append([
        r,
        accuracy_score(y, final_labels),
        recall_score(y, final_labels),
        precision_score(y, final_labels),
        f1_score(y, final_labels)
    ])

    all_true.extend(y)
    all_pred.extend(final_labels)

# ===================== RESULTS =====================
df = pd.DataFrame(
    rows, columns=["Record", "Accuracy", "Sensitivity", "PPV", "F1"]
)
print(df.sort_values("Record").to_string(index=False))

cm = confusion_matrix(all_true, all_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Non R-peak", "R-peak"]
)
disp.plot(cmap="Blues")
plt.title("Overall Confusion Matrix")
plt.show()

print("\nOVERALL:")
print("Accuracy   :", accuracy_score(all_true, all_pred))
print("Sensitivity:", recall_score(all_true, all_pred))
print("PPV        :", precision_score(all_true, all_pred))
print("F1-score   :", f1_score(all_true, all_pred))

# ===================== FINAL THESIS PLOT =====================
ecg, detected, annotated = get_final_peaks('104', rf)
plot_matched_detected_vs_annotated(ecg, detected, annotated)
```
